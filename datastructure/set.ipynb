```python
#set - set is an unordered collection of unique elements. 
#it does not allow duplicate value 

#unordered
a ={1,89,39,77,89,39,60}
print(a)
```

```python
#duplicates

a = {30,30,59,89,10,59}
print(a)
```

```python
#add

a = {"apple", "banana", "cherry"}

a.add("berry")
print(a)
```

```python
#remove
a = {"apple", "banana", "cherry","berry"}
a.remove("apple")
#a.remove("strawberry")
print(a)
```

```python
#discard

a = {"apple", "banana", "cherry","berry"}
#a.discard("apple")
a.discard("strawberry")
print(a)
```

```python
#union

a = {10,30,20,50,60}
b = {70,80,100,40,15}

#print(a.union(b))
print(a|b)
```

```python
#intersection - it returns common element
a = {10,30,100,50}
b = {70,30,100,20,15,50}

#print(a.intersection(b))
print(a&b)
```

```python
#difference - return element that are a but not in b

a = {1,2,3,4}
b = {3,4,5,6}

#print(a.difference(b))
print(a-b)
```

```python
#symmetric difference- returns element that are not common to both sets

a = {1,2,3,4}
b = {3,4,5,6}

#print(a.symmetric_difference(b))
print(a^b)
```

```python
#length
a = {1,2,3,4}
b = {3,4,5,6}
print(len(a))
print(len(b))
```

```python
#clear
a = {1,2,3,4}
b = {3,4,5,6}
print(a.clear())
#copy

c = b.copy()
print(c)
```

```python
#pop
a ={5,4,8,9,10}
a.pop()
print(a)
```

```python
#membership 

a = {10,20,30,40}
print(20 in a)
print(60 not in a)
```
