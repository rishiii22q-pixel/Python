```python
#A loop is a programming structure that repeatly execute a block of code as long as a condition or sequence required it

#types of loops
#for loop,while loop,nested loop

#for loop is used to iterate over the items for a sequence are other iterable such as list,tuple,dictionary,range,string
```

```python
#list
a = ['a', 'b','c','d']

for i in a:
    print(i)
```

```python
#tuple

sample =("Rishi","Saran","Abi")
for i in sample:
    print(sample)
```

```python
#set
sets = {"Rishi","Saran","Abi"}
for i in sets:
    print(sets)
```

```python
#string

a = "Rishi"
for i in a:
    print(i)
```

```python
dictionary ={"Rishi":1234,"Abishek":2311,"saran":93839}

for key,values in dictionary.items():
    print(key,values)

for key in dictionary:
    print(key)
```

```python
#Range

for i in range(1,10,2):
    print(i)
```

```python
#2table
for i in range(1,10+1):
    print(i,"x 2","=",2 *i)
    
```

```python

for i in range(1,10):
    print("* "*i,)
```

```python
#sum 
a = 0
for i in  range(1,6):
    a = a + i
print(a)
```

```python
#nested loop

for i in range(1,4):
    for j in range(1,4):
        print(j,i)
```

```python
#star 

for i in range(0,6):
    for j in range(i+1):
        print("*", end=" ")
    print()
```

```python
#nested
for i in range(1,9):
    for j in range(i):
        print(j+i,end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(i):
        print(i, end=" ")
    print()
```

```python
for i in range(1,5):
    for j in range(1,i+1):
        print(j, end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(5, 5-i,-1):
        print(j, end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(i):
        print(i-j,end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(i):
        print("*",end=" ")
    print()
for k in range(4,0,-1):
    for l in range(k):
        print("*", end=" ")
    print()
```
