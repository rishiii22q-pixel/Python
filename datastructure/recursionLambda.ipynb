```python
#factorial


def a(n):
    if n == 1:
     return 1
    return n * a(n-1)
print(a(6))
```

```python
#sum first n numbers

def num(n):
    if n == 1:
        return 1
    return n + num(n-1)
print(num(10))
```

```python
#lambda function is a small anonymous function that can have any number of parameters but only on expression

a = lambda num:num*num 
print(a(5))
```

```python
s = lambda a,b : a+b
print(s(5,5)) 
```

```python
s = lambda a,b : a  if a > b  else b
print(s(10,5))
```

```python
#map - The map() function applies a function to every element in an iterable (each as a list )
#syntax - map(function,iterable)

a = [1,2,3,4,5]
b = list(map(lambda x:x*x,a))
print(b)
```

```python

a = ['a','b','c','d','e']

b = list(map(lambda x:x.upper(),a))
print(b)
```

```python
#filter - filter() select only the element that satisfy the condition 

a = [1,2,3,4,5,6,7,8]
b = list(filter(lambda x : x%2 == 0,a))
print(b)
```

```python
a =[ 10,50,60,30,40,80]

b = list(filter(lambda x : x >=50, a))
print(b)
```

```python
#reduce - a reduce function combine all elements of an iteratable into a single value it is available in the functools modulus 
from functools import reduce

N =[10,40,50,20,80,12]

b=reduce(lambda a,b : a if a>b else b,N)
print(b)
```
