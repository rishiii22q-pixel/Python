```python
#dictionary - a dictionary is a collection of data stored as key value pairs.
#each key is unique and is used to access its corresponding value 
#stored data as key:values
#key must be unique
#values can be duplicated
#mutable can add,update,remove
#use {} curly braces

sample_dict ={"name":"rishi", "age":21, "course":"ds"}
print(sample_dict)


```

```python
#key and values
sample_dict ={"name":"rishi", "age":21, "course":"ds"}

print(sample_dict["age"])
```

```python
#get() - the get method returns the values of the specified key 
# if the key is not found it returns none inside open error

sample_dict ={"name":"rishi", "age":21, "course":"ds"}

print(sample_dict.get("name"))
```

```python
#key()
sample_dict ={"name":"rishi", "age":21, "course":"ds"}

print(sample_dict.keys())
print(sample_dict.values())
print(sample_dict.items())
```

```python
#update
sample_dict ={"name":"rishi", "age":21, "course":"ds"}

sample_dict.update({"age":27})
sample_dict.update({"city":"chennai"})
print(sample_dict)

```

```python
sample_dict ={"name":"rishi", "age":21, "course":"ds"}
sample_dict.pop("age")
print(sample_dict)
```

```python
#clear 
sample_dict ={"name":"abi", "city":"ALANDUR"}
sample_dict.clear()
print(sample_dict)
```

```python
#length
sample_dict ={"name":"rishi", "age":21, "course":"ds"}
print(len(sample_dict))
```

```python
#membership
sample_dict ={"name":"rishi", "age":21, "course":"ds"}

print("name" in sample_dict)
print("city" not in sample_dict)
```

```python
#copy
sample_dict ={"name":"rishi", "age":21, "course":"ds"}
b = sample_dict.copy()

print(b)
```
