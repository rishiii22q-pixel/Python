```python
#Read
f = open("sample.txt",'r')
d = f.read()
print(d)
f.close()
```

```python
#write 

n = open("sample.txt",'w')
n.write("Hiii PYTHON")
n.close()
```

```python
#append
n = open("sample.txt",'a')
n.write(" \n Hi Rishi! welcome to python programming")
n.close()
```

```python
#readline and readlines
n = open("sample.txt",'r')
d = n.readlines() #single line use n.readline()
print(d)
n.close()
```

```python
#create txt
n = open("sample1.txt",'x')
print(n.write("Hello Rishi!"))
n.close()
```
