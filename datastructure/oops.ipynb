```python
#constructor with parameters - we can pass values while creating an object

class student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

s1 = student("Rishi",21)

print(s1.name)
print(s1.age)
```

```python
class teacher():
    name ="siva"
    age = 25
    def studyTime(self):
        print("11.30 AM to 1.30 PM")

    def __init__(self,x,y):
        self.x = x
        self.y = y

t1 = teacher("Rishi",21)
print(t1.name)
print(t1.age)

t1.studyTime()
print(t1.x)
print(t1.y)


```

```python
#Encapsuation - it means wrapping data and method together inside a class and controlling access to the data (public,private,protected)
#Inheritance
#polymorphism
#abstraction
```

```python
#Encapsulation

"""  self.name - public
     self._name - protected
     self.__name - private
"""

class bankaccount:
    def __init__(self,balance):
        self.balance = balance
    def deposit(self,amount):
        self.balance = self.balance + amount
    def display(self):
        print("balance",self.balance)

b1 = bankaccount(10000)
print(b1.balance)
b1.deposit(1000)

print(b1.display())
```

```python
class student:
    def __init__(self):
        self.__name = "Rishi"

s1 = student()
s1._name ="saran"

print(s1.__name)
```

```python
#polymorphism

class dog:
    def sound(self):
        print("bark")
class cat:
    def sound(self):
        print("meow")

c = dog()
c.sound()
d = cat()
d.sound()
```

```python
#duck typing

class dog:
    def sound(self):
        print("bark")
class cat:
    def sound(self):
        print("mewo")
def make_sound(animal):
    animal.sound()
make_sound(dog())
make_sound(cat())
```

```python
#abstraction

#it means hiding unnecessary detailss and showing only the essential features

from abc import ABC, abstractmethod

class animal(ABC):
    @abstractmethod
    def sound(self):
        print("hello")

class dog:
    def sound(self):
        print("bark")
d = dog()
d.sound()
```

```python
# a class method work with the class rather than a particular instance

class student:
    collage = "ABC COLLAGE"
    @classmethod
    def change_collage(self,name):
        self.collage=name
student.change_collage("xyz collage")
print(student.collage)
```
