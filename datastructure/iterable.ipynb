```python
#an iterable is an python object that allows us to access its elements one by one

name = 'python'

for char in name:
    print(char)
```

```python
#iterator is an object that allows as you access element one at time

# __iter__()    __next__()  


n = [10,20,30]
my__iterator = iter(n)
print(next(my__iterator))
print(next(my__iterator))
print(next(my__iterator))

try:
    print(next(my__iterator))
except StopIteration:
    print("Program completed")

""" 
n is an iterable
iter(n) create an iterator
next() retrieves the next element

"""
```

```python
class STD:
    def __init__(self,limit):
        self.num = 1
        self.limit = limit
    def __iter__ (self):
        return self
    def __next__(self):
        if self.num <= self.limit:
            result = self.num
            self.num +=1
            return result
        else:
            raise StopIteration

n = STD(5)
for ch in n:
    print(ch)

  

    
```
