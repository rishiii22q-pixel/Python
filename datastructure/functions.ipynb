```python
def details(names,age):
    for i in range(len(names)):
        print("Name:", names[i])
        print("Age:", age[i])

names = ["Rishi","Saran","Praveen"]
age = [21,23,25]

details(names,age)
```

```python
def num(*details):
    for key,value in details.items():
        print(key,value)
num(name ="Rishi", age="21", city ="Chennai")
```
