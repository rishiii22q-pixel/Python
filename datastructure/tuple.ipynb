```python
#tuple

# A tuple is an ordered collection of multiple items stored in a single variable unlike a list at tuple is immuttable in which means its element cannot be changed 
#add or remove after creation 

a = (12,78,38)
```

```python
#indexing

a =(12,87,90)

print(a[1])
print(a[-1])
```

```python
#slicing
a = (10,20,30,40,50)

print(a[:3:1])

print(a[::-1])
```

```python
#count

a = (10,20,30,10,50,30)

print(a.count(10))
print(a.index(30))
```
