```python
#largest element in a list

numbers = [12, 45, 23, 67, 34, 89, 10]

i = 0
largest = numbers[0]

while i < len(numbers):
    if numbers[i] > largest:
        largest = numbers[i]
    i += 1

print("Largest:", largest)
```

```python
#Find Sum of List Elements

num=[10,20,30,40,50]
i = 0
total = 0 

while i < len(num):
    total += numbers[i]
    i+=1
print("sum:", total)

```

```python
#Count Even and Odd Numbers

numbers =[12,7,45,32,16,20]

i = 0
even = 0
odd = 0

while i < len(numbers):
    if numbers[i]%2==0:
        even = even + 1
    else:
        odd = odd + 1

    i = i + 1

print("EVEN NUMBERS COUNT:",even)
print("ODD NUMBERS COUNT:",odd)
```

```python
#Search a Element in List

List = [10,20,30,40,50,60]

target = 40
i =0
found = False

while i < len(List):
    if List[i] == target:
        found = True
        break
    i = i+1
if found:
    print("Element Found")
else:
    print("Element Not Found")
```

```python
#Remove duplicates from a list

numbers = [10, 20, 10, 30, 20, 40, 30]

unique = []
i = 0

while i < len(numbers):
    if numbers[i] not in unique:
        unique.append(numbers[i])
    i = i+ 1

print("Original:", numbers)
print("Unique:", unique)
```

```python
#Count Frequency using dictionary 

numbers = [10, 20, 10, 30, 20, 10]

frequency = {}
i = 0

while i < len(numbers):
    num = numbers[i]

    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

    i += 1

print(frequency)
```
