```python
#Student Marks Analyzer

student_marks = [98,54,77,86,45,63,72,34,20,39]

passed = 0
failed = 0

for i in range(len(student_marks)):
    mark = student_marks[i]

    if mark >= 90:
        grade = 'A'
    elif mark >= 75:
        grade = 'B'
    elif mark >= 50:
        grade = 'C'
    else:
        grade ='Fail'
    if mark >= 50:
        passed += 1
    else:
        failed += 1
    
    print(f"Student {i+1}: {mark} - Grade:{grade}")
    
print("***************************")
print("Total student passed:",passed)
print("Total student failed:",failed)

```

```python
#Number Analysis Program

""" Count the number of digits. 
2. Find the sum of all digits. 
3. Find the largest digit. 
4. Find the smallest digit. 
5. Count how many digits are even and odd. 
6. Display all the results."""

num = int(input("Enter the number:"))
digits = [int(i) for i in str(num)]

count = len(digits)
total_sum = sum(digits)
largest = max(digits)
smallest = min(digits)
even_count = sum(1 for i in digits if i%2 == 0)
odd_count = sum(1 for i in digits if i%2==1)


print("Number of digits:", count)
print("Sum Of digits:", total_sum)
print("Largest digit:", largest)
print("Smallest digit:", smallest)
print("Even digits count:", even_count)
print("Odd digit count:", odd_count)


```

```python
# Program to analyze digits of a number using while loop

num = int(input("Enter a number: "))
print(num)
temp = num

count = 0
total_sum = 0
largest = 0
smallest = 9
even_count = 0
odd_count = 0

while temp > 0:
    digit = temp % 10        # extract last digit
    count += 1               # count digits
    total_sum += digit       # add to sum
    
    if digit > largest:      # check largest
        largest = digit
    if digit < smallest:     # check smallest
        smallest = digit
    
    if digit % 2 == 0:       # even check
        even_count += 1
    else:                    # odd check
        odd_count += 1
    
    temp //= 10              # remove last digit

# Display results
print("Number of digits:", count)
print("Sum of digits:", total_sum)
print("Largest digit:", largest)
print("Smallest digit:", smallest)
print("Even digits count:", even_count)
print("Odd digits count:", odd_count)
```

```python
#Class Attendance Analyzer
    

student_attendance ={ "Arun":18,"Priya":22,"Kavin":15,"Divya":20,"Rahul":12}

total_working_day = 22
below = 0
#sample = list(student_attendance.values())
#print(sample)

for name, precent in student_attendance.items():
    percentage = (precent/total_working_day*(100))

    if percentage>=90:
        status = "Excellence"
    elif percentage>=75:
        status = "Good"
    elif percentage>=60:
        status = "Average"
    else:
        status = "Shortage"
    if percentage < 75:
        below = below + 1
        #name = (f"{name}-{percentage}-{status}")

    print(f"{name} - {percentage:.2f}% - {status}")

print("Students below 75:",below)
```

```python
# Multiplication Table Pattern

N = int(input("Enter N: "))

for i in range(1, N + 1):          # Outer loop for each table
    print(f"Table of {i}:")
    for j in range(1,N):         # Inner loop for 1 to 10
        result = i * j
        if result % 2 == 0:        # Check even/odd
            status = "Even"
        else:
            status = "Odd"
        print(f"{i} x {j} = {result}  {status}")
    print()  # Blank line between tables
```
