```python
#Employee salary Analyzer

salaries = [25000,45000, 32000,75000,18000,60000,52000 ]

def temp(salaries):
    high_salary = list(filter(lambda salary:salary>40000,salaries))
    average = sum(salaries)/len(salaries)

    highest = max(salaries)
    lowest = min(salaries)

    print("High salary Employee:",high_salary)
    print("Average salary:",average)
    print("Highest salary:",highest)
    print("Lowest salary:",lowest)


    for salary in salaries:

        if salary < 25000:
            print("Low:",salary)
        elif salary <= 50000:
            print("Medium:", salary)
        else:
            print("High:",salary)
    
temp(salaries)

```

```python
#Smart Password validator
def check_password(password):

    length = False
    uppercase =False
    lowercase = False
    digit = False
    special = False

    if len(password)>=8:
        length = True

    for char in password:
        if char.isupper():
            uppercase = True

        elif char.islower():
            lowercase = True

        elif char.isdigit():
            digit = True
        elif char in "@#$%":
            special = True

    check = lambda x : 1 if x else 0

    score =(check(length)
            +check(uppercase)
            +check(lowercase)+
            check(digit)+
            check(special))

    print("Password Length:","Valid" if length else "Invalid")
    print("Uppercase:","Present" if uppercase else "Missing")
    print("Lowecase:", "Present" if lowercase else "Missing")
    print("Digit:","Present" if digit else "Missing" )
    print("Special:","Prseent" if special else "Missing")

    if score == 5:
        print("Password Strength is High")
    elif score >=3:
        print("Password Strength is Medium")
    else:
        print("Password Strength is Weak")


password = input("Enter a password")

check_password(password)
```

```python
#Recursive number processor

def func(n):
    if n == 0:
        return 0,0,0

    digit = n % 10
    remaining = n // 10

    digit_sum , count, largest = func(remaining)

    digit_sum = digit_sum+digit
    count = count+1

    if digit > largest:
        largest = digit

    return digit_sum,count,largest

n = 583246

result = func(n)

print("sum of digits:",result[0])
print("number of digits:",result[1])
print("largest number:",result[2])
```

```python
#Transcation fraud detector

def detect(transaction):
    susp = []

    high_amount = lambda amount:amount > 50000
    for amount in transaction:

        if high_amount(amount) or amount < 0:
            print(amount,"suspicious")
            susp.append(amount)
        else:
            print(amount,"Normal")
    total = 0
    for amount in susp:
        total = total + amount

    print("suspicious transactions:", susp)
    print("total suspicious amount:", total)

transaction = [500, 1200, 75000, 300, 45000, 800, 150000, 2500]
detect(transaction)
```

```python
#recrusive palindrome checker

def is_palindrome(text):
    
    if len(text) <= 1:
        return True

    
    if text[0] != text[-1]:
        return False

    
    return is_palindrome(text[1:-1])


word = input("Enter word: ")

if is_palindrome(word):
    print(word, "is a Palindrome")
else:
    print(word, "is Not a Palindrome")
```
