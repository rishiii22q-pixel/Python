```python
#check key
student = {"Name":"Rishi","Age":21,"Course":"DS"}

Key = input("Enter the key value:")

if Key in student:
    print("Key Found")
else:
    print("Key Not Found")

```

```python
#remove duplicate values

list = [10,20,30,10,50]
print("entered values:", list)

print("after duplicates removed:", set(list))

```

```python
#Check element in set
student = {"Rishi", "Abi", "Priya","Rathish","Naveeth"}

Check = input("Enter the student Name:")

if Check in student:
    print("Student Attended")
else:
    print("Student Not Attended")
```

```python
#Set Operation

python = {"rishi", "abi","praveen","naveeth","thomas"}
java = {"rishi","praveen","Joel","Sakthi","Priya"}

print("Both clubs:", python & java)
print("Either club:", python | java)
```

```python
#Student Information
student = ("Arun",21,"EEE")

print("Name:", student[0])
print("Age:", student[1])
print("Dept", student[2])
```

```python
#access tuple elements
subjects = ("python","java","R","C","C++")
print("First Subject:", subjects[0])
print("Third Subject:", subjects[2])
print("Last Subject:", subjects[-1])
```

```python
#Tuple Slicing
numbers = (10,20,30,40,50,60)

print("First Three:", numbers[:3:])
print("Last Three:", numbers[3::])
```
