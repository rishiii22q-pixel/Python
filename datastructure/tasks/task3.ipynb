```python
#Student Result and Grade Calculator
Student_Name = input("Enter the Student Name:")

mark1 = int(input('Enter the Tamil Mark:'))
mark2 = int(input('Enter the English Mark:'))
mark3 = int(input('Enter the Maths Mark:'))
mark4 = int(input('Enter the Science Mark:'))
mark5 = int(input('Enter the Social Science Mark:'))

Average_Mark = (mark1+mark2+mark3+mark4+mark5)/5
print("AVERAGE MARK:",Average_Mark)

if Average_Mark >=90:
    print("Grade A")
if Average_Mark >= 75 and Average_Mark < 90:
    print("Grade B")
if Average_Mark >=60 and Average_Mark < 75:
    print("Grade C")
if Average_Mark >=50 and Average_Mark < 60:
    print("Grade D")
elif Average_Mark < 50:
    print("FAIL")

a = "FAILED THE EXAMINATION" if mark1 <35 or mark2 <35 or mark3 <35 or mark4 <35 or mark5 <35 else "PASSED THE EXAMINATION" 
print(Student_Name, a)



```

```python
#Employee Salary & Bonus Calculator

employee_details ={"EmployeeName":"RISHI", "Department":"IT","Basic_Salary":50000,"Experience":11}


if employee_details["Experience"] >= 10:
    bonus = 20
elif employee_details["Experience"]>=5:
    bonus = 15
elif employee_details["Experience"] >=2:
    bonus = 10
else:
    bonus = 5

if employee_details["Department"] == "IT" and employee_details["Experience"] > 5:
    bonus = bonus + 5


total = (employee_details["Basic_Salary"] * (bonus/100))
Total_Bonus = (total + employee_details["Basic_Salary"])

salary ="High Salary" if Total_Bonus>50000 else "Normal Salary"

print("Total Bonus Amount:",Total_Bonus)
print(salary)

```

```python
#Shopping Cart Discount System
items =["Laptop", "Mouse", "Keyboard", "Headset"]
Total_Amount = int(input("Enter the amount:"))

if Total_Amount >= 50000:
    discount = 20
elif Total_Amount >=25000:
    discount = 15
elif Total_Amount>=10000:
    discount = 10
else:
    discount =0

if "Laptop" in items and Total_Amount >30000:
    discount = discount+5

discount_amount = (discount/100 *Total_Amount)
final_amount = (Total_Amount - discount_amount)

delivery = "Free delivery" if final_amount >=5000 else "Delivery charge applicable"

print("Customer purchased items:", items)
print("Total Amount:", Total_Amount)
print("Discount:", discount)
print("Final Amount:" ,final_amount)
print("Delivery:",delivery)

```

```python
#Login and User Access System


login ={"username":"Admin", "password":"Admin@123","Role":"Manager"}

user =input("Enter the Username:")
if user in login["username"]:
    password = input("Enter the password:")
    if password in login["password"]:
        role = input("Enter your role:")
        if role in login["Role"]:
            print("ADMIN = ADMIN FULL ACCESS")
            print("MANAGER = MANAGER FULL ACCESS")
            print("STAFF = LIMITED ACCESS")
            print("Any other = NO ACCESS")
else:
    print("Login Fails")

login_status =("Login successful") if login["password"] == password else "Login Failed"
print(login_status)


```

```python
#Product Inventory Check
products =[
        {"Name": "Laptop","Price": 55000,"Stock": 10},
        {"Name": "Mouse","Price": 800,"Stock": 0},
        {"Name": "Keyboard","Price": 1500,"Stock": 5}]
index = int(input("Enter the product index value:"))

N = products[index]
print("Product:", N["Name"])
print("Product:",N["Price"])
print("Product:",N["Stock"])

if N["Stock"]==0:
    print("Out of stock")
elif N["Stock"]<=5:
    print("Low Stock")
else:
    print("Available")

if N["Price"]>50000 and N["Stock"]<5:
    print("High value Low Stock")

sale = "Available for sale" if N["Stock"] >0 else "Not Available"
print(sale)
        
        
```
