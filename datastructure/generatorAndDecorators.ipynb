```python
# a generator is a special type of iterator that produces values one at a time generator are created using: 1.generator function 2.generator expression

#example
def my_generator():
    yield 10
    yield 20
    yield 30
gen = my_generator()

print(next(gen))
print(next(gen))
print(next(gen))
""" 1.my_generator- is a generator function
    2. calling it create generator object
    3.next() - executes the function until the first yield
    4.the generator pauses at yield
    5.the next next() resumes execution from where it paused
    6.this process continues until the generator is exhausted """
```

```python
def count(n):
    for i in range(1,6):
        yield i
       

for num in count(5):
    print(num)



#1.return - return a value end the function - used in normal function , does not resume execution after a returning , a function normally return one final result
   #2.yield - produce a value and passes the function - used in generator function,can resume execution from where it passed, can produce a multiple values
    
```

```python
def count(n):
    for i in range(1,6):
        yield i**2
       

for num in count(5):
    print(num)
```

```python
def add():
    yield 10
    yield 20
    yield 30
for value in add():
    print(value)
```

```python
#expression is a compact way to create a generator 
#it looks similar to a list comprehension but it uses paranthesis instead of square bracket

#syntax: (expression for item in iterable)

s = (x for x in range(1,6+1))

for i in s:
   if i % 2 == 0:
      print(i)
```
