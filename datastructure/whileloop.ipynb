```python
#A while loop repeatedly executes a block of code as long as a condition is True


#while condition:
# statement
```

```python
i = 1
while i <=5:
    print(i)
    i = i +1
    #print(i)
  
```

```python
#list

list =["a","b","c","d"]

i = 0

while i < len(list):
    print(list[i], end =" ")
    i = i +1
```

```python
#String

string = "welcome"

i = 0

while i < len(string):
    print(string[i], end=" ")
    i = i+1
```

```python
#dictionary

dict = {"a":123,"b":67,"c":89}

keys = list(dict.keys())
i = 0
while i < len(keys):
    key = keys[i]
    print(key,":",dict[key])
    i = i+1
```

```python
student = {
    "name": "Arun",
    "age": 20,
    "city": "Dindigul"
}
keys = list(student.keys())
i = 0
while i < len(keys):
    key = keys[i]
    print(key, ":", student[key])
    i += 1
```

```python
#nested while

i =1 
while i <= 3:
    j = 1
    while j <=3:
        print(i,j)
        j = j+1
    i = i +1
```

```python
i = 1
while i <= 5:
    print("*"*i)
    i +=1
    
```
